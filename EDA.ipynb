```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("insurance_claim_fraud_dataset.csv")
```

```python
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="fraudulent"
)

plt.title("Fraudulent vs Non-Fraudulent Claims")
plt.xlabel("Fraudulent")
plt.ylabel("Number of Claims")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="claim_amount",
    bins=40,
    kde=True
)

plt.title("Distribution of Claim Amount")
plt.xlabel("Claim Amount")
plt.ylabel("Number of Claims")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="claim_status",
    hue="fraudulent"
)

plt.title("Fraudulent Claims by Claim Status")
plt.xlabel("Claim Status")
plt.ylabel("Number of Claims")
plt.legend(title="Fraudulent")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y="incident_type",
    hue="fraudulent"
)

plt.title("Fraudulent Claims by Incident Type")
plt.xlabel("Number of Claims")
plt.ylabel("Incident Type")
plt.legend(title="Fraudulent")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="claim_amount",
    y="payout_amount",
    hue="fraudulent",
    alpha=0.5
)

plt.title("Claim Amount vs Payout Amount")
plt.xlabel("Claim Amount")
plt.ylabel("Payout Amount")
plt.legend(title="Fraudulent")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="age",
    hue="fraudulent",
    bins=30,
    kde=True,
    element="step"
)

plt.title("Age Distribution by Fraud Status")
plt.xlabel("Age")
plt.ylabel("Number of Claims")

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="suspicious_documents",
    hue="fraudulent"
)

plt.title("Suspicious Documents vs Fraudulent Claims")
plt.xlabel("Suspicious Documents")
plt.ylabel("Number of Claims")
plt.legend(title="Fraudulent")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="delayed_reporting",
    hue="fraudulent"
)

plt.title("Delayed Reporting vs Fraudulent Claims")
plt.xlabel("Delayed Reporting")
plt.ylabel("Number of Claims")
plt.legend(title="Fraudulent")

plt.tight_layout()
plt.show()
```

```python
# Convert fraudulent column to numeric
df["fraudulent_numeric"] = df["fraudulent"].map({
    "Yes": 1,
    "No": 0
})

print(df[["fraudulent", "fraudulent_numeric"]].head())
```

```python
# Numerical columns
numeric_columns = [
    "claim_amount",
    "approved_amount",
    "deductible",
    "claim_duration_days",
    "age",
    "annual_income",
    "credit_score",
    "years_with_company",
    "previous_claims",
    "policy_duration_years",
    "premium_amount",
    "coverage_amount",
    "vehicle_year",
    "vehicle_age",
    "mileage",
    "vehicle_value",
    "repair_cost",
    "medical_cost",
    "towing_cost",
    "rental_vehicle_cost",
    "payout_amount",
    "fraudulent_numeric"
]

# Keep only columns that actually exist
available_columns = [
    col for col in numeric_columns
    if col in df.columns
]

print("Columns used for correlation:")
print(available_columns)

# Correlation
corr = df[available_columns].corr()

# Heatmap
plt.figure(figsize=(16, 12))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Insurance Claim Features")
plt.tight_layout()
plt.show()
```
